# Robot cell

Three robot arms you control by changing their state. The picture below each cell updates by itself.

Each arm has four values:

| Value | Options |
|---|---|
| `angle` | `0` facing you (the rack), `180` facing away (the conveyor) |
| `gripper_open` | `True` or `False` |
| `holding` | `True` or `False` |
| `raised` | `True` up, `False` down |

Keep this notebook in the same folder as `robot_cell.py` and `robot_cell_images/`.

```python
from robot_cell import RobotArm, RobotCell

cell = RobotCell(delay=0.5)
```

## Move one arm step by step

```python
arm = cell.arms[0]

arm.raised = False         # lower the arm to the rack
arm.gripper_open = False   # grab the tube
arm.holding = True
arm.raised = True          # lift it out
arm.angle = 180            # turn to face the conveyor
```

## Write your own arm class

Put the steps into methods, then give the class to `RobotCell` with `arm_class=`.

```python
class Picker(RobotArm):
    def pick_from_rack(self):
        self.angle = 0              # face the rack
        self.raised = False         # go down to the tube
        self.gripper_open = False   # grab it
        self.holding = True
        self.raised = True          # lift it out

    def place_on_conveyor(self):
        self.angle = 180            # face the conveyor
        self.raised = False         # lower it into the carrier
        self.gripper_open = True    # release the tube
        self.holding = False
        self.raised = True          # move clear


cell = RobotCell(delay=0.5, arm_class=Picker)
for arm in cell.arms:
    arm.pick_from_rack()
    arm.place_on_conveyor()
```
